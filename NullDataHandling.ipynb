```python
#df.show()
#df.printSchema()
#df.columns
from pyspark.sql.functions import sum, when, col

null_count_df = df.select([sum(when(col(c).isNull(), 1).otherwise(0)).alias(c) for c in df.columns])

display(null_count_df)
```
